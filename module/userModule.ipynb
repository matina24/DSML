```python
class Password:
    def __init__(self, password):
        self.password = password

    def check(self):
        if len(self.password) > 8 and self.password.isalnum():
            print("Strong Password")
        else:
            print("Weak Password")


# Object
p1 = Password("hello23456")
p1.check()
```

```python
numbers = [2, 3, 4, 5, 6, 7, 8, 9, 11, 12]

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

prime_numbers = list(filter(is_prime, numbers))

print(prime_numbers)
```

```python
test_numbers = [-10, -1, 0, 1, 2, 3, 4, 5, 6, 7, 9, 11, 13, 15, 17, 19, 21, 25, 27, 47, 49, 79, 97, 100, 523, 527, 1000, 7919, 10000, 104729, 1000000]

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_numbers = list(filter(is_prime, test_numbers))

print(prime_numbers)
```
