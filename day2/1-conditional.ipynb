```python
marks=int(input("Enter your marks: "))
if(marks>100 or marks<0):
    if marks>=80:
        print("Grade A")    
    elif marks>=60:
        print("Grade B")
    elif marks>=40:
        print("Grade C")
    else:
        print("Grade F")
else:
    print("Invalid marks")
```
