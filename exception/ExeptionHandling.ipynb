```python
class InvalidAgeException:
    """Raise when imput age is less than 18"""
    pass

def check_customer(age:int):
    if "liquior":
        if age >= 18:
            print("okay")
        else:
            raise InvalidAgeException("Must be an adult to buy liquior, age >18")
```

```python
try:
    check_customer(
        int(input("enter age"))
    )
except InvalidAgeException as e:
    print("Transaction Cancelled: Customer under 18")

except (ValueError,TypeError) as e:
    print("Invalid Age - Transaction failed")
```

```python
lst = ["a", "b", "c", "d"]

def get_element(index):
    return lst[index]

# get_element(0)
try:
    ui= int(input("enter index"))
    print(get_element(ui))
        
except IndexError as e:
     print("Index error, fall back to last element")
     print(get_element(-1))
    
except (ValueError,TypeError) as e:
    print("Invalid index ")
```

```python
#decorator

class Test:

    @staticmethod
    def something():
        print("Hello")


#implementation

def decorator(func): #-> function as argument/parameter

    #inner functiion
    def wrapper():
        print("before function call")
        print(func().upper()) # <- actual function call
        print("after function call")

    return wrapper # <- return function as value

@decorator
def greet():
    return "hello, world"


greet()
```
