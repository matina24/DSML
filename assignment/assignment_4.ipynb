Math Tutor Using Random, Arithmetic Operators, and OOP

```python
import random

class MathTutor:
    def __init__(self):
        self.correct = 0
        self.incorrect = 0

    def generate_question(self):
        num1 = random.randint(1, 20)
        num2 = random.randint(1, 20)
        operator = random.choice(["+", "-", "*", "/"])

        if operator == "+":
            answer = num1 + num2
        elif operator == "-":
            answer = num1 - num2
        elif operator == "*":
            answer = num1 * num2
        else:
            answer = round(num1 / num2, 2)

        return num1, num2, operator, answer

    def start(self):
        print("Welcome to Math Tutor!", flush=True)

        for i in range(5):
            num1, num2, operator, answer = self.generate_question()

            print(f"\nQuestion {i + 1}: {num1} {operator} {num2}", flush=True)

            try:
                user_answer = float(input("Your answer: "))

                if user_answer == answer:
                    print("Correct!")
                    self.correct += 1
                else:
                    print(f"Incorrect! Correct answer is {answer}")
                    self.incorrect += 1

            except ValueError:
                print("Invalid input! Please enter a number.")
                self.incorrect += 1

        self.show_score()

    def show_score(self):
        print("\n--- Final Score ---")
        print("Correct:", self.correct)
        print("Incorrect:", self.incorrect)
        print("Score:", self.correct, "/ 5")


tutor = MathTutor()
tutor.start()
```

Exception Handling Scenario: Online Age-Restricted Service

```python
class UnderageError(Exception):
    pass


def register_user(name, age):
    # Validate age
    if not isinstance(age, int):
        raise ValueError("Age must be a number.")

    # Check age restriction
    if age < 18:
        raise UnderageError("User must be 18 or older.")

    print(f"Welcome to MovieTime, {name}!")


try:
    name = input("Enter your name: ")
    age_input = input("Enter your age: ")

    # Convert age to integer
    try:
        age = int(age_input)
    except ValueError:
        raise ValueError("Age must be numeric.")

    register_user(name, age)

except UnderageError as e:
    print("Registration failed:", e)

except ValueError as e:
    print("Invalid input:", e)

else:
    print("Registration successful!")

finally:
    print("Thank you for using MovieTime!")
```

Online shopping cart system

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price


class User:
    def __init__(self, name, is_premium=False):
        self.name = name
        self.is_premium = is_premium
        self.cart = ShoppingCart(self)


class ShoppingCart:
    def __init__(self, user):
        self.user = user
        self.products = []

    def add_product(self, product):
        self.products.append(product)
        print(f"{product.name} added to cart.")

    def remove_product(self, product_name):
        for product in self.products:
            if product.name == product_name:
                self.products.remove(product)
                print(f"{product.name} removed from cart.")
                return

        print("Product not found.")

    def calculate_total_cost(self):
        total = sum(product.price for product in self.products)

        if self.user.is_premium:
            total = total * 0.90

        return total

    def generate_invoice(self):
        print("\n========== INVOICE ==========")
        print("Customer:", self.user.name)
        print("-----------------------------")

        subtotal = sum(product.price for product in self.products)

        for product in self.products:
            print(f"{product.name}: Rs. {product.price}")

        if self.user.is_premium:
            discount = subtotal * 0.10
        else:
            discount = 0

        total = subtotal - discount

        print("-----------------------------")
        print("Subtotal:", subtotal)
        print("Discount:", discount)
        print("Total:", total)
        print("=============================")


# Testing

product1 = Product("Laptop", 80000)
product2 = Product("Mouse", 1500)
product3 = Product("Keyboard", 3000)

user1 = User("Ram", is_premium=True)

user1.cart.add_product(product1)
user1.cart.add_product(product2)
user1.cart.add_product(product3)

print("\nTotal:", user1.cart.calculate_total_cost())

user1.cart.remove_product("Mouse")

print("\nTotal after removal:", user1.cart.calculate_total_cost())

user1.cart.generate_invoice()
```
