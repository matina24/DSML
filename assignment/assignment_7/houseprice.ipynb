```python
#Find outliers using IQR

Q1 = house_df[numerical_columns].quantile(0.25)
Q3 = house_df[numerical_columns].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = ((house_df[numerical_columns] < lower_bound) |
            (house_df[numerical_columns] > upper_bound))

outlier_count = outliers.sum()

outlier_count.sort_values(ascending=False)
```

Problem 2: Data Manipulation with Pandas

```python
# Calculate the 90th percentile of SalePrice
price_90th_percentile = house_df["SalePrice"].quantile(0.90)

print("90th Percentile of SalePrice:", price_90th_percentile)
```

```python
# Select houses with SalePrice above the 90th percentile

filtering_saleprice = house_df["SalePrice"].quantile(0.9)
high_price_df = house_df[house_df["SalePrice"] > filtering_saleprice]
high_price_df
```

```python
# Calculate price per square foot

house_df["Price_Per_SqFt"] = house_df.SalePrice / house_df.GrLivArea
house_df
```

```python
#Group by Neighborhood and calculate mean and median
neighborhood_prices = house_df.groupby("Neighborhood")["SalePrice"].agg(
    ["mean", "median"]
)

neighborhood_prices
```

```python
top_ten_price_per_sqft = house_df.sort_values(
    by="Price_Per_SqFt",
    ascending=False
).head(10)

top_ten_price_per_sqft
```

Problem 3: Handling Missing Data and Imputation

```python
for col in numerical_columns:
    house_df[col] = house_df[col].fillna(house_df[col].median())
```

```python
display(house_df[numerical_columns])
```

```python
for col in categorical_columns:
    house_df[col] = house_df[col].fillna(house_df[col].mode()[0])
```

Problem 4: Numpy Array Operations

```python
import numpy as np
```

```python
# Convert numerical columns to NumPy array
num_array = house_df[numerical_columns].to_numpy()

print("NumPy Array:")
num_array
```

```python
print("Shape of array:", num_array.shape)
```

```python
# Mean of each numerical column
column_mean = np.mean(num_array, axis=0)

# Standard deviation of each numerical column
column_std = np.std(num_array, axis=0)

print("Mean of each column:")
print(column_mean)

print("\nStandard deviation of each column:")
print(column_std)
```

```python
min_values = np.min(num_array, axis=0)
max_values = np.max(num_array, axis=0)

min_max_array = (
    (num_array - min_values) /
    (max_values - min_values)
)

print("Min-Max Scaled Array:")
print(min_max_array)
```

```python
z_score_array = (
    (num_array - column_mean) /
    column_std
)

print("\nZ-Score Normalized Array:")
print(z_score_array)
```

```python
grlivarea = house_df["GrLivArea"].to_numpy()
saleprice = house_df["SalePrice"].to_numpy()

dot_product = np.dot(grlivarea, saleprice)

print("\nDot Product:")
print(dot_product)
```

Problem 5: Data Visualization

```python
import seaborn as sns

plt.figure(figsize=(10, 6))

sns.histplot(house_df["SalePrice"], bins=30, kde=True)

plt.title("Distribution of SalePrice")
plt.xlabel("Sale Price")
plt.ylabel("Number of Houses")
plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.boxplot(
    data=house_df,
    x="OverallQual",
    y="SalePrice"
)

plt.title("SalePrice by Overall Quality")
plt.xlabel("Overall Quality")
plt.ylabel("Sale Price")
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=house_df,
    x="GrLivArea",
    y="SalePrice"
)

plt.title("GrLivArea vs SalePrice")
plt.xlabel("Above-Ground Living Area (sq ft)")
plt.ylabel("Sale Price")
plt.show()
```

```python
numerical_df = house_df.select_dtypes(
    include=["int64", "float64"]
)

correlation = numerical_df.corr()

plt.figure(figsize=(18, 14))

sns.heatmap(
    correlation,
    cmap="Purples",
    annot=True,       # Show correlation values
    fmt=".1f"         
)

plt.title("Correlation Heatmap of Numerical Features")
plt.show()
```
