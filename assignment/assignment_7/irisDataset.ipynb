3.Filtering and Sorting:
Filter the dataset to show only the rows where the sepal_length is greater than 5.0.
Sort the filtered dataset by petal_length in descending order.

```python
filtered_df = iris_df[iris_df["SepalLengthCm"] > 5.0]
filtered_df.head()
```

```python
sorted_df = filtered_df.sort_values("PetalLengthCm", ascending=False)
sorted_df.head()
```

6.Data Visualization:
Histogram:
Create a histogram to visualize the distribution of sepal_length values.
Scatter Plot:
Create a scatter plot with sepal_length on the x-axis and petal_length on the y-axis. Use different colors to represent different species.
Box Plot:
Create a box plot for the sepal_width values, grouped by species.

```python
import matplotlib.pyplot as plt

iris_df["SepalLengthCm"].plot(
    kind = "hist",
    bins = 10,
    edgecolor="black",
    xlabel = "Sepal length in cm",
    title="Distribution of Sepal Length"
)
```

The histogram shows how the sepal length values are distributed and how frequently different ranges occur.

```python
import seaborn as sns

sns.scatterplot(
    data = iris_df,
    x = "SepalLengthCm",
    y = "PetalLengthCm",
    hue="Species",
)
plt.xlabel("Sepal Length in cm")
plt.ylabel("Petal Length in cm")
plt.title("Sepal Length vs Petal Length")

plt.show()
```

The scatter plot shows the relationship between sepal length and petal length, with different colors representing the three species.

```python
sns.boxplot(
    data=iris_df,
    x="Species",
    y="SepalWidthCm"
)

plt.title("Sepal Width by Species")
plt.show()
```

The box plot compares the distribution of sepal width for each species, showing the median, spread, and possible outliers

7.Correlation Matrix
Calculate the correlation matrix for the numerical features and create a heatmap to visualize the correlations using seaborn.

```python
heat_map=iris_df.corr(numeric_only= True)
sns.heatmap(
    heat_map,
    annot= True,
    cmap="Purples",
)
```
