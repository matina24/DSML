Task 1:
Write a Python program using a lambda function to check if a given string is a palindrome

```python

text = input("Enter a string: ")

# Lambda function to check palindrome
check_palindrome = lambda s: s == s[::-1]

# Display the result
result = check_palindrome(text)

print(f" Is {text} palindrome:", result)
```

Task 2
The E-Commerce Analytics Pipeline (map, filter, reduce).
Using only map(), filter(), functools.reduce(), and lambda expressions (no explicit for or while loops):

```python
from functools import reduce

transactions = [
    {"id": 101, "amount": 250.0, "status": "completed", "currency": "USD"},
    {"id": 102, "amount": 0.0, "status": "pending", "currency": "USD"},
    {"id": 103, "amount": 180.5, "status": "completed", "currency": "USD"},
    {"id": 104, "amount": 45.0, "status": "refunded", "currency": "USD"},
    {"id": 105, "amount": 500.0, "status": "completed", "currency": "USD"},
]

```

```python
# 1. Filter only completed transactions

completed =list( filter(lambda transaction: transaction["status"] == "completed",transactions))
print(completed)
```

```python
# 2. Add 10% tax to each completed transaction

with_tax = list(map(lambda transaction: transaction["amount"] * 1.10,completed))
print(" Amounts after 10% tax:")
print(with_tax)
```

```python
# 3: Reduce values into total revenue

net_revenue = reduce(lambda total, amount: total + amount,with_tax,0)

print("Net Revenue:")
print(f"${net_revenue:.2f}")
```

Task 3:
Vehicle Rental Company

```python
from abc import ABC, abstractmethod


class Vehicle(ABC):

    def __init__(self, vehicle_id, brand, rental_price):
        self.__vehicle_id = vehicle_id
        self.__brand = brand
        self.__rental_price = rental_price


    def get_vehicle_id(self):
        return self.__vehicle_id

    def get_brand(self):
        return self.__brand

    def get_rental_price(self):
        return self.__rental_price

    
    def calculate_rental_cost(self, days):
        return self.__rental_price * days

    
    @abstractmethod
    def display_details(self):
        pass
```

```python
# Car class
class Car(Vehicle):

    def __init__(self, vehicle_id, brand, rental_price, number_of_doors):
        super().__init__(vehicle_id, brand, rental_price)
        self.__number_of_doors = number_of_doors

    def display_details(self):
        print("Car Details")
        print("-"*50)
        print("Vehicle ID:", self.get_vehicle_id())
        print("Brand:", self.get_brand())
        print("Rental Price per Day:", self.get_rental_price())
        print("Number of Doors:", self.__number_of_doors)
```

```python
# Bike class
class Bike(Vehicle):

    def __init__(self, vehicle_id, brand, rental_price, bike_type):
        super().__init__(vehicle_id, brand, rental_price)
        self.__bike_type = bike_type

    def display_details(self):
        print("Bike Details")
        print("-"*50)
        print("Vehicle ID:", self.get_vehicle_id())
        print("Brand:", self.get_brand())
        print("Rental Price per Day:", self.get_rental_price())
        print("Bike Type:", self.__bike_type)
```

```python
# Creating objects
car = Car(101, "Toyota", 50, 4)
bike = Bike(102, "Yamaha", 25, "Racing Bike")
```

```python
car.display_details()

print("Total rental cost for 3 days:", car.calculate_rental_cost(3))
```

```python
bike.display_details()

print("Total rental cost for 4 days:", bike.calculate_rental_cost(4))
```
