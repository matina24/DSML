Question: File Read and Write

Write a Program that Uses Functions write_to_file and read_from_file:

```python
def write_to_file(filename, content):
    with open(filename, "w") as file:
        file.write(content)


def read_from_file(filename):
    with open(filename, "r") as file:
        content = file.read()
        print(content)


# Write content to the file
write_to_file("greetings.txt", "Hello, Python!")

# Read content from the file
read_from_file("greetings.txt")
```

Task 3:
Number guessing game using random.randint()

```python
import random

while True:
    random_number = random.randint(1, 100)

    print("I have selected a number between 1 and 100.")
    print("You have 7 attempts to guess it.")

    for attempt in range(1, 8):
        guess = int(input("Enter your guess: "))

        if guess == random_number:
            print("Congratulations! You guessed the number!")
            break

        elif guess < random_number:
            print("Oh no!! The number is higher than that.")

        else:
            print("Oh no!! The number is lower than that.")

    else:
        print("Sorry! You used all 7 attempts.")
        print("The number was:", random_number)

    play_again = input("Would you like to play again? (yes/no): ")

    if play_again.lower() != "yes":
        print("Thanks for playing!")
        break
```

Task 4:
Checking if input number is armstrong or not

```python
number = int(input("Enter a positive integer: "))

digits = str(number)
number_of_digits = len(digits)

sum_of_powers = sum(int(digit) ** number_of_digits for digit in digits)

if sum_of_powers == number:
    print(f"Yes, {number} is an Armstrong number.")
else:
    print(f"No, {number} is not an Armstrong number.")
```
