Task 1:
Filtering given data using user defined functions and list comphrension.

```python
books = [
    ("The Alchemist", "Fiction", 1988, 250),
    ("The Da Vinci Code", "Mystery", 2003, 300),
    ("A Brief History of Time", "Science", 1988, 150),
    ("The Theory of Everything", "Science", 2002, 100),
    ("Pride and Prejudice", "Fiction", 1813, 200),
    ("To Kill a Mockingbird", "Fiction", 1960, 180),
    ("The Catcher in the Rye", "Fiction", 1991, 220),
    ("Angels & Demons", "Mystery", 2000, 210),
    ("The Grand Design", "Science", 2010, 90),
    ("1984", "Fiction", 1949, 190)
]

def filter_book(genre: str, year: int):
    y = [book[0] for book in books if book[1] == genre and book[2] >= year]

    print(y)


filter_book("Fiction", 1988)
```

Task 2:
Using lambda function to sort given list of books according to publication date

```python
books = [
    ("The Alchemist", "Fiction", 1988, 250),
    ("The Da Vinci Code", "Mystery", 2003, 300),
    ("A Brief History of Time", "Science", 1988, 150),
    ("The Theory of Everything", "Science", 2002, 100),
    ("Pride and Prejudice", "Fiction", 1813, 200),
    ("To Kill a Mockingbird", "Fiction", 1960, 180),
    ("The Catcher in the Rye", "Fiction", 1991, 220),
    ("Angels & Demons", "Mystery", 2000, 210),
    ("The Grand Design", "Science", 2010, 90),
    ("1984", "Fiction", 1949, 190)
]

x = [book[2] for book in books]

sort = lambda x: sorted(x)

x = sort(x)

y = [book for year in x for book in books if year == book[2]]

y
```

Task 3:
Number guessing game using random.randint()

```python
import random

random_number = random.randint(1, 100)

print("Welcome to the number guessing game.")
print("You get seven chances to guess a number between 1 and 100.")

conf = input("Do you want to play (y/n): ")

if conf == "y":
    for i in range(7):
        user_guess = int(input("Enter your guess between 1 and 100: "))

        if user_guess == random_number:
            print("Wow!! You guessed the number correctly!")
            break

        elif user_guess > random_number:
            print("Oh no!! The number is lower than that.")

        else:
            print("Oh no!! The number is higher than that.")

    else:
        print("Sorry! You used all seven chances.")
        print("The correct number was:", random_number)

    print("Thank you for participating!")
```

Task 4:
Checking if input number is armstrong or not

String input version

```python
num = input("Enter a positive number: ")
num=num.strip("-")

s = 0

for i in num:
    s = s + int(i) ** len(num)

if str(s) == num:
    print(f"{num} is an Armstrong number")
else:
    print(f"{num} is not an Armstrong number")
```

Integer input version

```python
num = int(input("Enter a positive number: "))

s = 0

for i in str(num):
    s = int(i) ** len(str(num)) + s

if s == num:
    print(f"{num} is an Armstrong number")
else:
    print(f"{num} is not an Armstrong number")
```
