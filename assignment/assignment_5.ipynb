TASK 2:
Creating a NumPy Array:

```python
# 1. Create a 1D NumPy array from a Python list

arr1 = np.array([1, 2, 3, 4, 5])
print("1D Array:")
print(arr1)
```

```python
# 2. Create a 2D NumPy array of shape (3x3) using numbers 1 to 9
arr2 = np.random.randint(1,9 ,size = (3, 3))
print("2D Array:")
print(arr2)
```

```python
# 3. Generate 10 evenly spaced values between 0 and 5
arr3 = np.linspace(0, 5,num= 10)
print("Evenly Spaced Array:")
print(arr3)
```

TASK 3:
Indexing and Slicing Arrays:

```python
print("Original Array:")
print(arr2)

# 1. Access element in second row, third column

element = arr2[1:2, 2:]
print("\nElement at second row, third column:", element)
```

```python
# 2. Slice first two rows and first two columns

sliced_array = arr2[:2, :2]
print("First two rows and first two columns:")
print(sliced_array)
```

```python
# 3. Modify last row, first column to 100

arr2[2, 0] = 100
print("\nModified Array:")
print(arr2)
```

TASK 4:
Properties and Methods of NumPy Arrays

```python
# 1. Find shape, size, and data type


print("Shape:", arr2.shape)
print("Size:", arr2.size)
print("Data Type:", arr2.dtype)
```

```python
# 2. Change 1D array into 2D array of shape (5,1)

print(arr1)

arr1_2d = arr1.reshape(5, 1)

print("1D array reshaped to (5,1):")
print(arr1_2d)
```

```python
# 3. Flatten multi-dimensional array into 1D array


flattened = arr2.flatten()

print("Flattened array:",flattened)
```

TASK 5 :
Operations on NumPy Arrays

```python
# 1. Add 5 to every element in the 1D array

result1 = np.add(arr1,5)
print("1D array after adding 5:", result1)
```

```python
# 2. Multiply every element of the 2D array by 3

result2 = np.multiply(arr2,3)
print("2D array after multiplying by 3:")
print(result2)
```

```python
# 3. Matrix multiplication

A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

result3 = np.dot(A, B)

print("Matrix Multiplication:")
print(result3)
```

TASK 6:
Understanding Broadcasting

```python
# Create a 3x3 matrix of ones
matrix = np.ones((3, 3))

# Create a 1D array of length 3
arr = np.array([1, 2, 3])
```

```python
# Add the 1D array to each row using broadcasting

result = matrix + arr

print("Matrix of ones:")
print(matrix)

print("\n1D Array:")
print(arr)

print("\nResult after broadcasting:")
print(result)
```
