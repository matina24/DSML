```python
class dog:
    count = 0

    def __init__(self, name: str, age: int = 1):
        #initializer or constructor
        self.name= name # instance attribute
        self.age = age # instance attribute

        dog.count +=1
```

```python
class person:
    count = 0

    def __init__(self, name: str, address: str = "kathmandu"):
        #initializer or constructor
        self.name= name # instance attribute
        self.address = address # instance attribute

        person.count +=1

    def get_details(self): # instance method
        print(f"I am {self.name} from {self.address}")

    @classmethod
    def get_count(cls):
        return cls.count

    @staticmethod
    def get_full_name(first, last): #static mathod
        return f"{first},{last}"
```

```python
student1 = person("john","norway")
student1.get_details()
student1.get_count()
```

```python
student1.get_full_name("brandon", "heffies")
```
