```python
class Animal:
    def __init__(self,name, age):
        self.name= name
        self.age = age

    def intro(self):
        print(f"hello, I am {self.name}")
```

```python
class Cat(Animal):
    def __init__(self, name1,age1,color):
        super().__init__(name1,age1)
        self.color = color

    def details(self):
        print(f"{self.color} Cat:\n{self.name} \nAge: {self.age}")
```

```python
cat = Cat("Luffy",2,"White")
cat.intro()
```

```python
cat.details()
```

Multiple Inheritance

```python
class parent1:
    def func1(self):
        print("This is parent1")

class parent2:
    def func2(self):
        print("This is parent2")

class parent3:
    def func1(self):
        print("This is parent3")
```

```python
class Child(parent1, parent2, parent3):
    def func4(self):
        print("This is child function")
```

```python
obj= Child()
obj.func4()
```

```python
obj.func1()
```

```python
class person():
    def __init__(self, name:str, age: int):
        self.name=name
        self.age= age


class employee(person):
    def __init__(self,name,age, em_id):
         super().__init__(name, age)
         self.em_id= em_id

class Manager(employee):
    def __init__(self, name, age, em_id, department):
        super().__init__(name, age,em_id)
        self.department= department

    
    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Employee ID:", self.em_id)
        print("Department:", self.department)
        


manager = Manager("Bhawana", 25, "E101", "IT")
manager.display()

```

```python
class Animal:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def intro(self):
        print(f"hello, I am {self.name}")


class Mammal(Animal):
    def __init__(self, name, age, fur=False):
        super().__init__(name, age)
        self.fur = fur

    def birth(self):
        print(f"{self.name} gives birth to live ones")


class Bird(Animal):
    def fly(self):
        return f"{self.name} can fly"


class Bat(Mammal, Bird):
    pass


b = Bat("chamero", 4)

b.intro()
b.birth()
print(b.fly())
```
