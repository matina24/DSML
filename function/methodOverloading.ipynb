Encapsulation

```python
class Example:
    def __init__(self):
        self.public=1
        self.__private=23
        self._protected=45

    def public_method(self):
        print("Public Method")

    def __private_method(self):
        print("Private Method")

    def _protected_method(self):
        print("protected method")


obj= Example()
obj.public  #yes
obj._protected  #yes
obj.__private  #no
```

```python
obj._protected_method()
```

```python
class SunClass(Example):
    def get_public(self):
        print(self.public)

    def get_protected(self):
            print(self._protected)

    def get_private(self):
            print(self.__private)

sun_obj = SunClass()
sun_obj.get_public()
```

```python

sun_obj.get_protected()
```
