```python
class Person:
    """
    Hello
    """
    def __init__(self,name, age):
        self.name= name
        self.age= age

    def __str__(self):
        return f"{self.name}/ {self.age}"

    def __repr__(self):
        return f"{self.name} is {self.age}"

    def __len__(self):
        return self.age

    def __call__(self, gender="female"):
        self.gender=gender
        return f"{self.name} is {self.gender}"
        

person = Person("Tisa", 23)
print(person)

        
```

```python
class Point:
    def __init__(self,x,y):
        self.x= x
        self.y = y

    def __str__(self):
        return  f"{self.x},{self.y}"

    def __add__(self,other):
        new_x=self.x + other.x
        new_y=self.y + other.y
        return Point(new_x, new_y)

    def __mul__(self, other):
        new_x=self.x * other.x
        new_y=self.y * other.y
        return Point(new_x, new_y)
        


```

```python
p1= Point(5,9)
p2 = Point(5,1)

p3= p1 + p2
print(p3)
```

```python
pm= p1 * p2
print(pm)
```

```python
class quad_region:

    def __init__(self, x, y):
        self.x = x
        self.y = y

    def quadrant(self):
        if self.x > 0 and self.y > 0:
            print("First Quadrant")
        elif self.x < 0 and self.y > 0:
            print("Second Quadrant")
        elif self.x < 0 and self.y < 0:
            print("Third Quadrant")
        elif self.x > 0 and self.y < 0:
            print("Fourth Quadrant")
        elif self.x == 0 and self.y == 0:
            print("Origin")
        else:
            print("Point lies on an axis")

qr= quad_region(5,9)
qr.quadrant()
```

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def quadrant(self):
        if self.x==0 and self.y==0:
            print("origin")
        else:
            signs = (self.x > 0, self.y > 0)

            if signs == (True, True):
                print("First Quadrant")
            elif signs == (False, True):
                print("Second Quadrant")
            elif signs == (False, False):
                print("Third Quadrant")
            else:
                print("Fourth Quadrant")


p = Point(0,0)
p.quadrant()
```
